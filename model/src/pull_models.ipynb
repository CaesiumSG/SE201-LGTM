# Pull models

Downloads every model in `MODELS` (from `model/.env`) into scratch, so serving jobs start without a download. Needs no GPU; `model/deploy/provision.sh` runs it as a CPU job.

```python
from ollama_server import OllamaServer, load_config

config = load_config()
models = config.get("MODELS", config["DEFAULT_MODEL"]).split()

server = OllamaServer(config)
server.start()
for model in models:
    server.pull(model)
print("available:", server.models())
server.stop()
```
